```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


aq = pd.read_csv("E:\\PROJECT FILE\\PROJECT 8-Anomaly Detection in Air Quality Data for Early Identification of Pollution Events\\datasets\\air quality data\\city_day.csv")
weather = pd.read_csv("E:\\PROJECT FILE\\PROJECT 8-Anomaly Detection in Air Quality Data for Early Identification of Pollution Events\\datasets\\india weather data\\Weather_data.csv")
fires = pd.read_csv("E:\\PROJECT FILE\\PROJECT 8-Anomaly Detection in Air Quality Data for Early Identification of Pollution Events\\datasets\\viirs-snpp_2024_India.csv")

```

```python
# parse datetime
aq["Datetime"] = pd.to_datetime(aq["Datetime"])

# pick one city
city_name = "Delhi"   # change to a city that exists in your data
aq_city = aq[aq["City"] == city_name].sort_values("Datetime")

# ensure daily frequency (city_day is already daily, so just set index)
aq_city = aq_city.set_index("Datetime").asfreq("D")

aq_city.head()
```

```python
# convert acq_date string to datetime
fires["acq_date"] = pd.to_datetime(fires["acq_date"])

# daily number of VIIRS detections in India
fires_daily = fires.groupby("acq_date").size().reset_index(name="fire_count_all_india")

fires_daily.head()
```

```python
# bring air data index back as a normal column for merge
data = aq_city.reset_index().rename(columns={"Datetime": "date"})

data = pd.merge(
    data,
    fires_daily,
    left_on="date",
    right_on="acq_date",
    how="left"
)

# days with no fires get NaN -> treat as zero fires
data["fire_count_all_india"] = data["fire_count_all_india"].fillna(0)

# set date as index again
data = data.set_index("date")

data[["City", "PM2.5", "AQI", "fire_count_all_india"]].head()
```

```python
# Simple fill for missing pollutant/AQI values
pollutant_cols = ["PM2.5","PM10","NO","NO2","NOx","NH3","CO","SO2","O3","Benzene","Toluene","Xylene","AQI"]

data[pollutant_cols] = data[pollutant_cols].interpolate(limit_direction="both")

```

```python
pm_col = "PM2.5"

mu = data[pm_col].mean()
sigma = data[pm_col].std()

data["z_score_pm25"] = (data[pm_col] - mu) / sigma
data["is_anomaly_pm25"] = data["z_score_pm25"].abs() > 3  # 3 std dev threshold

data[[pm_col, "z_score_pm25", "is_anomaly_pm25", "fire_count_all_india"]].head()
```

```python
plt.figure(figsize=(14,5))

plt.plot(data.index, data[pm_col], label="PM2.5", color="blue")

# plot anomalies as red dots
plt.scatter(
    data.index[data["is_anomaly_pm25"]],
    data.loc[data["is_anomaly_pm25"], pm_col],
    color="red",
    label="Anomalies (|z|>3)"
)

plt.xlabel("Date")
plt.ylabel("PM2.5 concentration")
plt.title(f"{city_name} PM2.5 with simple z-score anomalies")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
feature_cols = [
    "PM2.5",
    "PM10",
    "NO2",
    "SO2",
    "CO",
    "O3",
    "AQI",
    "fire_count_all_india"
]

X = data[feature_cols].copy()
X = X.interpolate(limit_direction="both")
X = X.fillna(0)
X
```

```python
from sklearn.ensemble import IsolationForest
```

```python
# tell the model: roughly what fraction of days you expect to be weird
contamination = 0.03  # 3% anomalies, you can tweak this

iso = IsolationForest(
    n_estimators=200,      # number of trees
    contamination=contamination,
    random_state=42
)

iso.fit(X)
```

```python
# predictions: -1 = anomaly, 1 = normal
pred = iso.predict(X)

# anomaly score (higher = more abnormal)
scores = iso.decision_function(X)  # actually: lower score = more abnormal in sklearn

data["iforest_score"] = -scores          # flip so bigger = more abnormal (easier to think)
data["iforest_label"] = (pred == -1)     # True = anomaly

data[["AQI", "PM2.5", "fire_count_all_india", "iforest_score", "iforest_label"]].head()
```

```python
data.columns
```

```python
plt.figure(figsize=(14,5))

plt.plot(data.index, data["AQI"], label="AQI", color="blue")

plt.scatter(
    data.index[data["iforest_label"]],
    data.loc[data["iforest_label"], "AQI"],
    color="red",
    label="IsolationForest anomalies"
)

plt.xlabel("Date")
plt.ylabel("AQI")
plt.title(f"{city_name} AQI with IsolationForest anomalies")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# read weather data if not already loaded
weather = pd.read_csv("E:\\PROJECT FILE\\PROJECT 8-Anomaly Detection in Air Quality Data for Early Identification of Pollution Events\\datasets\\india weather data\\Weather_data.csv")

# 1) convert epoch time to real datetime
weather["last_updated_epoch"] = pd.to_datetime(weather["last_updated_epoch"], unit="s")

# 2) keep only date (no time) for grouping
weather["date"] = weather["last_updated_epoch"].dt.date

# 3) create one row per day with average weather
weather_daily = (
    weather
    .groupby("date")[["temperature_celsius", "wind_kph", "humidity"]]
    .mean()
    .reset_index()
)

weather_daily.head()
```

```python
# ensure 'date' is a datetime and also make a pure date for matching
# if your date column is called 'Datetime'
data = data.rename(columns={"Datetime": "date"})
```
